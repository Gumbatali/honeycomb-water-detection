```python
from analyze_water_cooling import load_video_memmap

video, fps = load_video_memmap(Path("data/water/water1.mat"), "data")
fps = fps if fps > 0 else 30

fig1_sl = [210, 70, 295, 155]
fig2_sl = [300, 70, 385, 155]

def crop(img, bbox):
    x1, y1, x2, y2 = bbox
    return img[y1:y2 + 1, x1:x2 + 1]

def to_u8(frame):
    low, high = np.nanpercentile(frame, [5, 95])
    norm = np.clip((frame - low) / max(high - low, 1e-6), 0, 1)
    return (norm * 255).astype(np.uint8)

out_path = "outputs/mix_fig1_fig2.mp4"
writer = cv2.VideoWriter(
    out_path,
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (86 * 3, 86),
    isColor=False,
)

for i in range(video.shape[2]):
    frame = np.asarray(video[:, :, i], dtype=np.float32)
    frame_u8 = to_u8(frame)

    fig1 = crop(frame_u8, fig1_sl)
    fig2 = crop(frame_u8, fig2_sl)

    mixed = cv2.addWeighted(fig1, 0.7, fig2, 0.3, 0)
    mixed = np.hstack((fig1, fig2, mixed))
    writer.write(mixed)

writer.release()
print(out_path)
```
